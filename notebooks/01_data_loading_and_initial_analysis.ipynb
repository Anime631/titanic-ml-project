```python
import pandas as pd
import numpy as np
import os

DATA_DIR = "../data/raw/"

train_df = pd.read_csv(os.path.join(DATA_DIR, "train.csv"))
test_df  = pd.read_csv(os.path.join(DATA_DIR, "test.csv"))

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)

train_df.head()
```

```python

print("=== TRAIN INFO ===")
train_df.info()

print("\n=== TEST INFO ===")
test_df.info()
```

```python

print("=== TRAIN: Missing ===")
m_train = train_df.isnull().sum()
p_train = (m_train / len(train_df) * 100).round(2)
print(pd.DataFrame({"missing": m_train, "pct": p_train}))

print("\n=== TEST: Missing ===")
m_test = test_df.isnull().sum()
p_test = (m_test / len(test_df) * 100).round(2)
print(pd.DataFrame({"missing": m_test, "pct": p_test}))
```

```python
print("=== TRAIN describe() ===")
print(train_df.describe())

print("\n=== TEST describe() ===")
print(test_df.describe())
```

```python
for col in ["Pclass","Sex","Embarked","Survived"]:
    print(f"\n=== {col} ===")
    print(train_df[col].value_counts(dropna=False))
```

# Day 1 — Data Loading and Initial Analysis

## Goal
Perform a first-look audit of the Titanic dataset before any modeling.

## Data
- `train.csv` — 891 rows, 12 columns (includes target `Survived`)
- `test.csv` — 418 rows, 11 columns (no target)

## What We Checked
1. **Shape** — train (891, 12), test (418, 11)
2. **Dtypes & non-null counts** — `.info()`
3. **Missing values** — counts + percentages
4. **Numeric statistics** — `.describe()`
5. **Categorical counts** — `.value_counts(dropna=False)`

## Key Findings
- **Missing values (train):**
  - `Age` — ~20% missing
  - `Cabin` — ~77% missing
  - `Embarked` — ~0.22% missing (2 rows)
- **Target balance:** 61.6% died, 38.4% survived (imbalanced)
- **Key categories:**
  - `Sex` — 577 male, 314 female
  - `Pclass` — 491 in 3rd, 216 in 1st, 184 in 2nd
  - `Embarked` — 644 from S, 168 from C, 77 from Q

## Implications for Later Days
- `Age` needs imputation (Day 3).
- `Cabin` is too sparse → transform to binary "HasCabin" or drop (Day 3).
- Class imbalance → track precision/recall, not just accuracy (Day 5).
